```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from matplotlib.colors import ListedColormap
```

```python
amostras_imgs = []
amostras_lbs = []

caminhos = ['../datasets/graphene/graphene_dataset0.npz', '../datasets/graphene_defects/graphene_defects_dataset0.npz','../datasets/mos2/mos2_dataset_0.npz', '../datasets/mos2_defects/mos2_dataset_defects_0.npz', '../datasets/mos2_S_defects/mos2_dataset_S_defects_2.npz']

for i in range(4):

    dataset = np.load(caminhos[i])
    images = dataset['images']
    labels = dataset['labels']

    _, images = train_test_split(
        images, test_size=0.2, random_state=42)
    
    _, labels = train_test_split(
        labels, test_size=0.2, random_state=42)


    img = images[0]
    lb = labels[0]

    amostras_imgs.append(img)
    amostras_lbs.append(lb)

i=4
dataset = np.load(caminhos[i])
images = dataset['images']
labels = dataset['labels']

_, images = train_test_split(
    images, test_size=0.2, random_state=42)

_, labels = train_test_split(
    labels, test_size=0.2, random_state=42)

img = images[11]
lb = labels[11]

amostras_imgs.append(img)
amostras_lbs.append(lb)
```

```python
fig, axes = plt.subplots(2, 5, figsize=(20, 8))

fig.suptitle("Amostras de imagens e máscaras por dataset", fontsize=24)

cmap2 = ListedColormap(["blue", "red"])
cmap3 = ListedColormap(["blue", "red", "lightgray"])
cmap4 = ListedColormap(["blue", "red", "yellow", "lightgray"])

titulos = ["Grafeno", "Defeitos de Grafeno", "MoS$_2$", "Defeitos de MoS$_2$", "Defeitos de Enxofre em MoS$_2$"]
cmaps = [cmap2, cmap2, cmap3, cmap3, cmap4]

labels = ["Imagem Convoluída", "MaxPooling", "FuzzyPooling"]

axis_y = ["Imagem", "Máscara"]

for i, label in enumerate(axis_y):
    axes[i, 0].text(
        -0.1, 0.5, label,
        transform=axes[i, 0].transAxes,
        rotation=90,
        va="center",
        ha="center",
        fontsize=18
    )

for i in range(5):
    axes[0,i].imshow(amostras_imgs[i][0], cmap='gray')
    axes[1,i].imshow(amostras_lbs[i][0], cmap=cmaps[i])
    axes[0,i].set_title(titulos[i], fontsize=18)
    axes[0,i].axis('off'); axes[1,i].axis('off')
```
