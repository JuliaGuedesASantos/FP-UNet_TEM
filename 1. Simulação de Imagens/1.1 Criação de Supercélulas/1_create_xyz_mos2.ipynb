```python
import ase
from ase.io import read
from ase.visualize import view
from ase.build import make_supercell

import numpy as np
# from atoms_mod import orthogonalize_cell, write_xyz # pedir pra ana esse arquivo depois
```

### Criando supercélula de MoS2

```python
atoms = ase.build.mx2(vacuum=2)

mos2 = atoms.copy()

# Matrix to transform the 120-degree hexagonal cell into a 90-degree rectangular cell
transformation_matrix = [[1, 0, 0], 
                         [1, 2, 0], 
                         [0, 0, 1]]
 
mos2_ortho = make_supercell(mos2, transformation_matrix)
```

```python
view(mos2_ortho)
```

```python
mos2_ortho *= (40, 20, 1)
view(mos2_ortho)
```

```python
natoms = len(mos2_ortho)
defect_percent = 10/100
ndefects = int(np.rint(defect_percent*natoms))
natoms,ndefects

defects_index = np.random.choice(np.arange(natoms),size=ndefects,replace=False)

defects_pos = mos2_ortho[defects_index]

view(defects_pos)

atoms_defects = mos2_ortho.copy()
del atoms_defects[defects_index]

view(atoms_defects)

defects_pos.write('mos2_defects_pos.poscar')
atoms_defects.write('mos2_defects.poscar')
```

```python
#write_xyz(mos2_defects, 'mos2_computem.xyz')
with open('mos2_computem.xyz','w') as f:
    f.write('this line is comment\n')
    f.write(f'    {atoms_defects.cell.array[0,0]} {atoms_defects.cell.array[1,1]} {atoms_defects.cell.array[2,2]}\n')
    zvals = atoms_defects.get_atomic_numbers()
    positions = atoms_defects.get_positions()
    for io in range(len(atoms_defects)):
        f.write(f'   {zvals[io]:>2d}  {positions[io,0]:>9.6f}  {positions[io,1]:>9.6f}  {positions[io,2]:>9.6f}  1.0\n')
    f.write('   -1')
```
