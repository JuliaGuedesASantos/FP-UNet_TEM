```python
import numpy as np
import matplotlib.pyplot as plt
from ase import build
from ase.visualize import view
```

```python
atoms = build.graphene_nanoribbon(n=40,m=40,vacuum=10.0,sheet=True) # cria o sistema de grafeno, com célula unitária de 2 átomos. n e m multiplica a célula unitária em x e y
```

```python
view(atoms)
```

```python
# vira a célula para sair do eixo yz para xy

cell_array = atoms.cell.array
positions_array = atoms.get_positions()

print(cell_array)
print(positions_array)

new_cell_array = cell_array.copy()
new_cell_array[2,2] = cell_array[1,1]
new_cell_array[1,1] = cell_array[2,2]

new_positions = positions_array.copy()
new_positions[:,1] = positions_array[:,2]
new_positions[:,2] = positions_array[:,1]

graphene = atoms.copy()
graphene.set_cell(new_cell_array)
graphene.set_positions(new_positions)
```

```python
#inserção de defeitos

natoms = len(graphene)
defect_percent = 10/100
ndefects = int(np.rint(defect_percent*natoms))
natoms,ndefects
```

```python
defects_index = np.random.choice(np.arange(natoms),size=ndefects,replace=False)
```

```python
defects_pos = graphene[defects_index]
```

```python
defects_pos.write('graphene_defects_pos.poscar')
```

```python
view(defects_pos)
```

```python
graphene_defects = graphene.copy()
del graphene_defects[defects_index]
```

```python
view(graphene_defects)
```

```python
graphene_defects.write('graphene_defects.poscar')
```

```python
with open('graphene_computem.xyz','w') as f:
    f.write('this line is comment\n')
    f.write(f'    {graphene_defects.cell.array[0,0]} {graphene_defects.cell.array[1,1]} {graphene_defects.cell.array[2,2]}\n')
    zvals = graphene_defects.get_atomic_numbers()
    positions = graphene_defects.get_positions()
    for io in range(len(graphene_defects)):
        f.write(f'   {zvals[io]:>2d}  {positions[io,0]:>9.6f}  {positions[io,1]:>9.6f}  {positions[io,2]:>9.6f}  1.0\n')
    f.write('   -1')
```
